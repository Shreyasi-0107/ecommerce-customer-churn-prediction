# Business Insights and Recommendations

The customer churn prediction model helps identify customers who are likely to stop purchasing from the platform.

By analyzing customer purchasing behavior, engagement patterns, and historical transactions, the model provides insights into the factors that influence customer retention and churn.

The following business insights were identified from Exploratory Data Analysis, model feature importance, and SHAP explainability analysis.


## 1. Customer Recency and Engagement

Customers with a longer gap since their last purchase show a higher likelihood of churn.

A decrease in recent activity indicates reduced customer engagement and can serve as an early warning signal.

### Recommendation:
- Launch targeted re-engagement campaigns.
- Send personalized discount offers to inactive customers.
- Use reminder notifications to encourage repeat purchases.

## 2. Purchase Frequency and Loyalty

Customers with fewer completed orders generally show weaker engagement compared to frequent buyers.

Repeat purchasing behavior is a strong indicator of customer loyalty.

### Recommendation:
- Introduce loyalty reward programs.
- Provide incentives for repeat purchases.
- Encourage customers to return through personalized recommendations.

## 3. Customer Spending Patterns

Customer spending patterns help distinguish between high-value customers and occasional buyers.

High-value customers contribute significantly to revenue and require proactive retention strategies.

### Recommendation:
- Create VIP customer segments.
- Provide exclusive offers for valuable customers.
- Prioritize retention efforts for high-revenue customers.

## 4. Customer Tenure

Customer tenure provides information about the relationship duration between customers and the platform.

New customers require stronger engagement strategies to convert them into long-term customers.

### Recommendation:
- Improve onboarding experience.
- Provide first-purchase incentives.
- Engage new customers during the early lifecycle stage.

# Project Conclusion

This project developed an end-to-end customer churn prediction system using the Olist Brazilian E-commerce dataset.

The complete workflow included:

- Data cleaning and preprocessing
- Feature engineering from customer transaction history
- Exploratory data analysis to understand customer behavior
- Development and evaluation of machine learning models
- Model interpretation using SHAP explainability techniques

Multiple machine learning algorithms were evaluated, including Logistic Regression, Random Forest, and XGBoost.

Among the evaluated models, XGBoost achieved the best performance and was selected as the final prediction model.

The SHAP analysis improved model transparency by identifying the key factors influencing churn predictions.

The final solution enables businesses to identify customers at risk of leaving and take proactive retention actions.

# Future Work

Although the current churn prediction system provides valuable insights, several improvements can enhance its performance and business usability.

Future improvements include:

- Hyperparameter optimization using GridSearchCV, RandomizedSearchCV, or Optuna.
- Implementing time-based validation to better simulate real-world prediction scenarios.
- Incorporating additional features such as product categories, customer reviews, and seller information.
- Developing customer segmentation models for personalized marketing strategies.
- Creating an interactive dashboard using Streamlit.
- Deploying the model as an API for real-time churn prediction.
- Monitoring model performance after deployment to detect changes in customer behavior.
